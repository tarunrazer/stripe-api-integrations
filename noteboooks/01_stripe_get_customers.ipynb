```python
import os
import httpx

# Read the key securely from environment memory
STRIPE_API_KEY = os.getenv("STRIPE_TEST_KEY", "sk_test_placeholder")

headers = {
    "Authorization": f"Bearer {STRIPE_API_KEY}"
}

def fetch_stripe_customers():
    # 1. API Contract Details (validated first in Postman)    
    url = "https://api.stripe.com/v1/customers"
    params = {"limit": 2}

    # 2. Make Network Request via httpx
    response = httpx.get(url, headers=headers, params=params)

    # 3. Defensive Status Handling & Payload Normalization
    if response.status_code == 200:
        raw_json = response.json()
        customer_data = raw_json.get("data", [])

        # Clean mapping into structured dicts
        parsed_customers = []
        for cust in customer_data:
            parsed_customers.append(
                {
                    "customer_id": cust.get("id"),
                    "email": cust.get("email", "N/A"),
                    "created_timestamp": cust.get("created"),
                    "currency": cust.get("currency", "usd"),
                }
            )
        return {
            "status": "SUCCESS",
            "count": len(parsed_customers),
            "customers": parsed_customers,
        }
    else:
        return {"status": "FAILED", "http_code": response.status_code}


# Execute function and print output
result = fetch_stripe_customers()
print("Stripe API Customer Retrieval Output:")
print(result)
```
