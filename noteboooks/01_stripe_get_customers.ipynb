```python
import os
import httpx
from dotenv import load_dotenv

# 1. Load key-value pairs from local .env into system memory
load_dotenv(dotenv_path="../.env")

# 2. Retrieve key securely
api_key = os.getenv("STRIPE_TEST_KEY")


def fetch_stripe_customers():
    url = "https://api.stripe.com/v1/customers"
    headers = {"Authorization": f"Bearer {api_key}"}
    params = {"limit": 3}

    response = httpx.get(url, headers=headers, params=params)

    if response.status_code == 200:
        data = response.json()
        customers = data.get("data", [])
        return [
            {
                "id": c.get("id"),
                "email": c.get("email", "N/A"),
                "created": c.get("created"),
            }
            for c in customers
        ]
    else:
        return {"error": response.status_code, "detail": response.text}


# Execute
result = fetch_stripe_customers()
print("Secure Stripe Customer Fetch Output:")
print(result)
```
